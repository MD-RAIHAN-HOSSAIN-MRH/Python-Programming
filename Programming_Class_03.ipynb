```python
# while loop to print all the numbers upto 500,
# for every number, you have to output the type of the number,
# positive and even
# positive and odd,
# negative and even,
# negative and odd,

count = 0

while(count <= 500):
  if(count > 0 and count%2 == 0):
    print(f"{count} is Positive and Even")
  elif(count > 0 and count%2 != 0):
    print(f"{count} is Positive and Odd")
  elif(count < 0 and count%2 == 0):
    print(f"{count} is Negative and Even")
  else:
    print(f"{count} is Negative and Odd")

  count = count + 1

```

```python
# searhcing in a list
X = ["potato","tomato","onion","shampoo","carrot"]

target = "tomato"
# setting a flag
# flaging

# flag = False
flag = 0
for item in X:
  if(item == target):
    flag = flag + 1
    # flag = True

if(flag != 0):
  print("Item has been found!")
else:
  print("Item has not been found!")

print(f'{target} has been found {flag} times in the list!')
```
