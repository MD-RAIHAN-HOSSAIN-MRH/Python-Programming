```python
# # Binary Search

def binary_search(arr,target):
  start = 0
  end = len(arr) - 1
  print("This is the start index --> " + str(start))
  print("This is the end index --> " + str(end))
  while(start <= end):
    mid = (start + end)//2
    if(arr[mid] > target):
      end = mid - 1
    elif(arr[mid] < target):
      start = mid + 1
    else:
      print("This is the start value --> " + str(start))
      print("This is the end value --> " + str(end))
      return True

  print("This is the start value --> " + str(start))
  print("This is the end value --> " + str(end))
  return False


binary_search([10,20,30,40,50,60,70,80],35)
```

```python
# Sorting Algorithms

# Bubble Sort
# Ascending Order

arr = [40,34,54,80,15]

for j in range(len(arr) - 1):
  for i in range(len(arr) - 1):
    if(arr[i] > arr[i+1]):
      # make the swap
      arr[i],arr[i+1] = arr[i+1],arr[i]
      print(arr)


print("This is the final array: ")
print(arr)
```

```python
# Bubble Sort
# Descending Order

arr = [40,34,54,80,15]

for j in range(len(arr) - 1):
  for i in range(len(arr) - 1):
    if(arr[i] < arr[i+1]):
      # make the swap
      arr[i],arr[i+1] = arr[i+1],arr[i]
      print(arr)


print("This is the final array: ")
print(arr)
```

```python
# Selection Sort
# Ascending order

arr = [7,4,5,6]

for i in range(len(arr) - 1):
  minimum_index = i
  for j in range(i+1, len(arr)):
    if(arr[j] < arr[minimum_index]):
      minimum_index = j
  arr[i], arr[minimum_index] = arr[minimum_index], arr[i]
  print(arr)

print("This is the final list:")
print(arr)
```

```python
# Selection Sort
# Descending order

arr = [7,4,5,6]

for i in range(len(arr) - 1):
  minimum_index = i
  for j in range(i+1, len(arr)):
    if(arr[j] > arr[minimum_index]):
      minimum_index = j
  arr[i], arr[minimum_index] = arr[minimum_index], arr[i]
  print(arr)

print("This is the final list:")
print(arr)
```

```python
# Insertion Sort
# Ascending Order

arr = [7,4,5,6]

for i in range(len(arr)):
  for j in range(i):
    if(arr[j] > arr[i]):
      # make the swap here!
      arr[i], arr[j] = arr[j], arr[i]
    print(arr)

print(arr)
```

```python
# Insertion Sort
# Descending Order

arr = [7,4,5,6]

for i in range(len(arr)):
  for j in range(i):
    if(arr[j] < arr[i]):
      # make the swap here!
      arr[i], arr[j] = arr[j], arr[i]
    print(arr)

print(arr)
```
