```python
# Linked List

class Node:
  def __init__(self,data,next):
    self.data = data
    self.next = next

N1 = Node(10,None)
N2 = Node(20,None)
N3 = Node(30, None)
N4 = Node(40, None)
N5 = Node(50, None)

# Make the change here for making the linked list:
N1.next = N2
N2.next = N3
N3.next = N4
N4.next = N5

#testing the linked list
print(N2)
print(N1.next)
print(N1.data)
print(N2.data)
print(N5.next)
```

```python
# Iterating over the linked list:
# Find out the size of the linked list
iterator = N1
values = ""
count = 0
while(iterator.next != None):
  value = iterator.data
  values = values + str(value) + " ----> "
  # This is the code for jumping from one node to another!
  iterator = iterator.next
  count = count + 1

print(values)
print(f"This is the size of the linked list: {count + 1}")
```

```python
# Using OOP to make a linked list class

# Helper Class
class Node:
  def __init__(self,data,next):
    self.data = data
    self.next = next


class LinkedList:
  def __init__(self):
    self.head = None

  def insert_at_beginning(self,value):
    new_node = Node(value,self.head)
    self.head = new_node

  def print_list(self):
    iterator = self.head
    linkedlist = ""
    while(iterator.next != None):
      linkedlist = linkedlist + str(iterator.data) + " ----> "
      # This is the code for jumping from one node to another!
      iterator = iterator.next

    linkedlist = linkedlist + str(iterator.data) + "----> None"
    print(linkedlist)

  def insert_at_end(self,value):
    iterator = self.head
    while(iterator.next != None):
      # This is the code for jumping from one node to another!
      iterator = iterator.next

    new_node = Node(value, None)
    iterator.next = new_node

  def count(self):
    iterator = self.head
    count = 0
    while(iterator.next != None):
      # This is the code for jumping from one node to another!
      iterator = iterator.next
      count = count + 1

    print(f"This is the size of the linked list: {count + 1}")

  def insert_at_index(self,value,index):
    if(index == 0):
      self.insert_at_beginning(value)

    iterator = self.head
    count = 0
    while(iterator):
      iterator = iterator.next
      count =count + 1
      if(count == index - 1):
        new_node = Node(value,iterator.next)
        iterator.next = new_node

  def delete_at_index(self,index):
    # index == 0
    if(index == 0):
      self.head = self.head.next

    # index != 0
    count = 0
    iterator = self.head

    while(iterator):
      if (count == index - 1):
        iterator.next = iterator.next.next

      iterator = iterator.next
      count = count + 1



L1 = LinkedList()
L1.insert_at_beginning(10)
L1.insert_at_beginning(20)
L1.insert_at_beginning(30)
L1.insert_at_beginning(40)

L1.print_list()

L1.insert_at_end(5)
L1.insert_at_end(2)

L1.print_list()

L1.count()

L1.insert_at_index(25,2)
L1.print_list()

L1.delete_at_index(0)

L1.print_list()
```
