```python
# function sum() that finds out the total of all the students in  the class
Marks = [85,80,70]


def sum_of_class(Marks):
  sum = 0
  for  i in Marks:
    sum  = sum + i
  print(f" The sum of all the marks in the class is {sum}")


sum_of_class(Marks)
```

```python
# function avg() that finds out the total of all the students in  the class
Marks = [85,80,70]

def avg_of_class(Marks):
  sum = 0
  for  i in Marks:
    sum  = sum + i

  average = sum / len(Marks)
  print(f" The average of all the marks in the class is {average}")

avg_of_class(Marks)
```

```python
# 60 and up --> Good Performance
# else --> Needs Improvement

def remarks_on_class_performance(Marks):
  sum = 0
  for  i in Marks:
    sum  = sum + i

  average = sum / len(Marks)
  print(f" The average of all the marks in the class is {average}")

  if( average >= 60):
    print("Good Performance")
  else:
    print("Nees Improvement!")

remarks_on_class_performance(Marks)
```

```python
# Take a list containg numbers and then double every single value in  the list
# [10,20,30] --> [20,40,60]


Marks = [10,20,30]

def double_list():
  for i in range(len(Marks)):
    # print(i)
    #print(Marks[i])
    Marks[i] = Marks[i] * 2

print(Marks)
double_list()
print(Marks)
```

```python
# You will be taking 2 lists containg numbers,
# find out which list has a sum that is larger

list_01 = [10,20,30,35243]
list_02 = [20,40,60]

def compare_lists():
  sum_01 = 0
  for i in list_01:
    sum_01 = sum_01 + i

  sum_02 = 0
  for i in list_02:
    sum_02 = sum_02 + i

  if(sum_01 > sum_02):
    print("The first list is bigger!")

  else:
    print('The second list is bigger')

compare_lists()
```
