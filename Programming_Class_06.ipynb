```python
# student class

class student:

  def __init__(self,name,student_id,course_name,marks):
    # This is an attribute
    self.name = name
    self.student_id = student_id
    self.course_name = course_name
    self.marks = marks

  # Method!
  def greet(self):
    print("Hello, I am in class now!")

  def name_your_course(self):
    print(f"the name of my course is {self.course_name}")

  def tell_your_name(self):
    print(f"my name is {self.name}")

  def change_marks(self):
    new_mark = int(input("Enter the new mark/ updated mark for the student: "))
    self.marks = new_mark

  def comment_on_marks(self):
    if(self.marks >= 60):
      print("Pass!")
    else:
      print("Fail!")

Student01 = student("Alex",123,"Graphics Design",85)
Student02 = student('Sam',124,"Blockchain Technology",78)
Student03 = student("Smith", 125,"Video Editing",67)
Student04 = student("Jones",126,"Digital Marketing",65)
Student05 = student("Stacey",127,"Web Design",20)

Student01.name_your_course()

print(Student01.marks)
Student01.change_marks()
print(Student01.marks)

Student01.comment_on_marks()
```
