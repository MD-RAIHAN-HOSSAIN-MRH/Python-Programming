```python
# Take a number from  the user
# then ask the user for that many inputs
# add each input in a list called  --> Numbers

Numbers = []

times = int(input("How many numbers do you want to insert?: ")) # 5

for i in range(times):
  value = int(input("Give me the value to enter: "))
  Numbers.append(value)

print(Numbers)
```

```python
# You are given a list
# find out how many even numbers are there in  the list

Numbers = [1,2,3,4,5,6,7,8,9]

count = 0
for item in Numbers:
  if( item % 2 == 0):
    count = count + 1

print(count)
```

```python
# You are given a list
# find out how many positive numbers are there in  the list

Numbers = [-1,2,-3,4,5,6,-7,-8,-9]

count = 0
for item in Numbers:
  if(item > 0):
    count = count + 1

print(count)
```

```python
# sort out the Numbers list into even and odd lists

Numbers = [1,2,3,4,5,6,7,8,9]
even_list = []
odd_list = []

for item in Numbers:
  if(item % 2 ==0):
    even_list.append(item)
  else:
    odd_list.append(item)

print(even_list)
print(odd_list)
```

```python
# Iterating over a dictionary

Marks = {
    "Sam" : 100,
    "Alex": 98,
    "Tom": 85
}

# Getting the keys in  the dicionary
print("These are all the keys: ")
for i in Marks:
  print(i)

# iterating over all the values
print('These are all the values: ')
for i in Marks:
  print(Marks[i])
```

```python
# ways to iterate over a dictionary
Marks = {'Alex': 80, 'Tom': 50, 'Jane': 90}

for i in Marks:
  print(i)

print("------------------------------")

for i in Marks:
  print(Marks[i])


print("------------------------------")

for i in Marks.keys():
  print(i)

print("------------------------------")

for i in Marks.values():
  print(i)

print("------------------------------")

for i in Marks.items():
  print(i)
```

```python
# Task
# make 2 list
# 1 list should contain all the keys
# another list should contain all the values


Marks = {'Alex': 80, 'Tom': 50, 'Jane': 90}

all_keys = []
all_values= []

for i in Marks.keys():
  all_keys.append(i)

for i in Marks.values():
  all_values.append(i)


print(all_keys)
print(all_values)
```
