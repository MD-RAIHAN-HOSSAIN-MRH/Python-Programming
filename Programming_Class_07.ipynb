```python
# More on OOP

class teacher:

  def __init__(self,name,teacher_id,courses_taught,enrolled_students):
    self.name = name
    self.teacher_id = teacher_id
    self.courses_taught = courses_taught
    self.enrolled_students = enrolled_students
  def add_new_students(self,name):
    self.enrolled_students.append(name)
  def find_student(self,name):
    if(name in self.enrolled_students):
      print("True")
    else:
      print("False")
  def query_course(self,course_name):
    if(course_name in self.courses_taught):
      print("Found!")
    else:
      print("Not Found!")

Teacher01 = teacher("Mr. Fahim",123,["Blockchain Technology","Data Science","Quantum Computing"],["Arshad","Fatema"])
Teacher01.find_student("Alex")
```

```python
# Class Activity 02

class Vehicle:

  def __init__(self,max_speed,color,number_of_seats):
    self.max_speed = max_speed
    self.color = color
    self.number_of_seats = number_of_seats
    self.max_acceptable_speed =  100

  def check_speed(self):
    if(self.max_speed > self.max_acceptable_speed):
      print("Speed is too high!")
    else:
      print('Acceptable speed')

  def change_color(self,new_color):
    self.color = new_color

  def change_max_speed(self,speed):
    self.max_speed = speed

  def calculate_least_time(self,distance):
    time = distance/self.max_speed
    print(time)

  def calculate_acceptable_least_time(self,distance):
    time = distance/self.max_acceptable_speed
    print(time)

Car01 = Vehicle(120,"Green",4)
print(Car01.max_acceptable_speed)
Car01.check_speed()
```
