```python
# length --> Taken from the user
# area --> Taken from  the user
# You have to determine if the object is a square of a rectangle

length = int(input("Give me the lenght of the figure: "))
area = int(input("Give me the area of the figure: "))

calculated_area = length * length

if(area == calculated_area):
  print("The figure is a square!")
else:
  print("The  figure is a Rectangle!")
```

```python
# User will give you a number and then you have to decide if the number is:
# positive and even
# positive and odd
# negative and even
# negative and odd

number = int(input("Input the number: "))

if(number > 0 and number%2 == 0):
  print("Positive and Even")

elif(number > 0 and number%2 != 0):
  print("Positive and Odd")

elif(number < 0 and number%2 == 0):
  print("Negative and Even")

else:
  print("Negative and Odd")
```

```python
# Take a number from  the user and then
# print all the even numbers from 0 till that number!

number = int(input("Give me a number: ")) # ----> 10

count = 0
while(count <= number):
  if(count % 2 == 0):
    print(count)
  count = count + 1
```

```python
# Take a number from  the user and then
# print all the odd numbers from 0 till that number!

number = int(input("Give me a number: ")) # ----> 10

count = 0
while(count <= number):
  if(count % 2 != 0):
    print(count)
  count = count + 1
```
