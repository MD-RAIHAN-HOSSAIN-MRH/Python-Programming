```python
# implementing the stack using OOP

class Stack:
  def __init__(self):
    self.stack = []

  def push(self,value):
    self.stack.append(value)

  def pop(self):
    value = self.stack.pop()
    print(f"This is the value that was popped ---> {value}")

  def peek(self):
    print(self.stack[-1])

  def size_of(self):
    print(f'This is the size of the stack ----> {len(self.stack)}')

  def is_stack_empty(self):
    if(len(self.stack) == 0):
      print('The stack is empty')
    else:
      print("The stack is not empty")

stack01 = Stack()
stack02 = Stack()
stack03 = Stack()
stack04 = Stack()

print(stack01.stack)
```

```python

class Queue:
  def __init__(self):
    self.queue = []

  def enqueue(self,value):
    self.queue.append(value)

  def dequeue(self):
    self.queue.pop(0)

  def size_of(self):
    print(f"This is the size of the queue ---> {len(self.queue)}")

  def peek(self):
    print(self.queue[0])

  def is_queue_empty(self):
    if(len(self.queue) == 0):
      print("The queue is empty!")
    else:
      print("The queue is not empty!")


Queue01 = Queue()
Queue01.enqueue(100)
Queue01.enqueue(200)
Queue01.enqueue(300)
Queue01.enqueue(400)
```
